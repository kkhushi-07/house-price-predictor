```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#read the data which is in csv format
data = pd.read_csv('C:\\Users\\user\\OneDrive\\Desktop\\house-price-predictor\\data\\california_housing.csv')
```

```python
data
```

```python
data.info()
```

```python
#drop the rows with missing values
data.dropna(inplace=True)
```

```python
#now there are no missing values in the dataset
data.info()
```

```python
from sklearn.model_selection import train_test_split#in this line we are inporting the train_test_split function from the sklearn library which is used for splitting the dataset into training and testing sets.
x=data.drop('median_house_value', axis=1)#remove the target variable from the dataset and store the remaining features in a new variable called x.
y=data['median_house_value']#store the target variable in a new variable called y.
```

```python
x
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
train_data = x_train.join(y_train)
```

```python
train_data
```

```python
train_data.hist(figsize=(15,8))
```

```python
plt.figure(figsize=(15, 8))
#it shows the correlation between the features in the dataset. The correlation values range from -1 to 1, where -1 indicates a strong negative correlation, 0 indicates no correlation, and 1 indicates a strong positive correlation.
sns.heatmap(train_data.corr(numeric_only=True), annot=True, cmap='YlGnBu')#numeric_only=True is used to include only numeric columns in the correlation calculation. annot=True is used to display the correlation values on the heatmap. cmap='YlGnBu' is used to specify the color map for the heatmap.
```

```python
#at frist these four features are highly skewed and they are not normally distributed so we will apply log transformation to these features to make them normally distributed.
train_data['total_rooms']=np.log(train_data['total_rooms']+1)
train_data['total_bedrooms']=np.log(train_data['total_bedrooms']+1)
train_data['population']=np.log(train_data['population']+1)
train_data['households']=np.log(train_data['households']+1)
```

```python
train_data.hist(figsize=(15,8))
```

```python
#one-hot encoding is a technique used to convert categorical variables into numerical format. In this case, the ocean_proximity column is a categorical variable that needs to be converted into numerical format for machine learning algorithms to work with it. 
train_data.ocean_proximity.value_counts()
```

```python
#The pd.get_dummies() function is used to create new columns for each unique value in the ocean_proximity column, with a value of 1 indicating the presence of that category and a value of 0 indicating its absence. The resulting DataFrame is then joined with the original train_data DataFrame using the join() method, and the original ocean_proximity column is dropped using the drop() method.
train_data.join(pd.get_dummies(train_data.ocean_proximity))
```

```python
#one-hot encoding -> pd.get_dummies() function is used for attach new 5 columns to the train_data DataFrame, each representing a unique value in the ocean_proximity column. The resulting DataFrame is then joined with the original train_data DataFrame using the join() method, and the original ocean_proximity column is dropped using the drop() method.
train_data=train_data.join(pd.get_dummies(train_data.ocean_proximity)).drop(['ocean_proximity'], axis=1)
```
